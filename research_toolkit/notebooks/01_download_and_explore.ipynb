# Notebook 1: Download and Explore FinanceBench

**Objective**: Download FinanceBench dataset and perform exploratory data analysis.

**Outputs**:
- `data/raw/financebench_open_source.jsonl`
- EDA visualizations
- Summary statistics

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datasets import load_dataset

# Add src to path
sys.path.insert(0, str(Path.cwd().parent / "src"))

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Download FinanceBench from HuggingFace

```python
print("📥 Downloading FinanceBench from HuggingFace...")
fb_dataset = load_dataset("PatronusAI/financebench")

# Convert to DataFrame
fb_df = fb_dataset['train'].to_pandas()

print(f"✅ Loaded {len(fb_df)} questions")
print(f"Columns: {fb_df.columns.tolist()}")
```

```python
# Save to data/raw/
output_path = Path('../data/raw/financebench_open_source.jsonl')
output_path.parent.mkdir(parents=True, exist_ok=True)

fb_df.to_json(output_path, orient='records', lines=True)
print(f"💾 Saved to {output_path}")
```

## 2. Exploratory Data Analysis

```python
# Basic info
print("Dataset shape:", fb_df.shape)
print("\nFirst row:")
fb_df.iloc[0]
```

```python
# Missing values
print("Missing values:")
fb_df.isnull().sum()
```

### Company Distribution

```python
company_counts = fb_df['company'].value_counts()

plt.figure(figsize=(14, 8))
company_counts.head(20).plot(kind='barh')
plt.title('Top 20 Companies in FinanceBench', fontsize=14)
plt.xlabel('Number of Questions')
plt.ylabel('Company')
plt.tight_layout()
plt.show()

print(f"\nTotal unique companies: {fb_df['company'].nunique()}")
print(f"Questions per company (mean): {company_counts.mean():.1f}")
```

### Question Length Distribution

```python
fb_df['question_length'] = fb_df['question'].str.len()
fb_df['answer_length'] = fb_df['answer'].str.len()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Question length
sns.histplot(fb_df['question_length'], bins=30, ax=axes[0])
axes[0].set_title('Question Length Distribution')
axes[0].set_xlabel('Characters')

# Answer length
sns.histplot(fb_df['answer_length'], bins=30, ax=axes[1])
axes[1].set_title('Answer Length Distribution')
axes[1].set_xlabel('Characters')

plt.tight_layout()
plt.show()

print(f"Question length (mean): {fb_df['question_length'].mean():.0f} chars")
print(f"Answer length (mean): {fb_df['answer_length'].mean():.0f} chars")
```

### Sample Questions and Answers

```python
print("\n📋 Sample Questions:\n")
print("=" * 80)

for i, row in fb_df.sample(5, random_state=42).iterrows():
    print(f"\n{i+1}. Company: {row['company']}")
    print(f"   Q: {row['question']}")
    print(f"   A: {row['answer']}")
    print("-" * 80)
```

### Document Distribution

```python
if 'doc_name' in fb_df.columns:
    doc_counts = fb_df['doc_name'].value_counts()
    
    print(f"Unique documents: {fb_df['doc_name'].nunique()}")
    print(f"\nTop 10 documents:")
    print(doc_counts.head(10))
```
