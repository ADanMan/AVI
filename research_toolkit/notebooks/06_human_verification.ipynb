**Objective**: Manual verification of experiment results

```python
import sys
from pathlib import Path
import pandas as pd

sys.path.insert(0, str(Path.cwd().parent / 'src'))

from experiment.human_review import HumanReview
```

## 1. Load Results

```python
results_dir = Path('../data/results')
latest_exp = sorted(results_dir.glob('experiment_*'))[-1]
results_df = pd.read_csv(latest_exp / 'raw_results.csv')
print(f'✅ Loaded {len(results_df)} results')
```

## 2. Identify Cases for Review

```python
review = HumanReview(results_df)

cases = review.identify_review_cases(
    flags=['low_confidence_judge', 'metric_disagreement', 'random_sample']
)

print(f'✅ Identified {len(cases)} cases for review ({len(cases)/len(results_df):.1%})')
```

## 3. Export for Annotation

```python
output_path = latest_exp / 'human_review.csv'
review.export_for_review(str(output_path), cases=cases)
```
